## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*
This dataset has some limitations and cannot provide complete information about all aspects of content performance.

1. Unbalanced history:
The dataset may contain different amounts of historical data for different pages or clients. Some content may have longer performance histories while others may have limited records, so comparisons may not always be fair.

2. GSC-only early rows:
Some early records may only contain Google Search Console (GSC) data and may not include GA4 analytics data. Therefore, we cannot fully understand user engagement or traffic behavior for those periods.

3. Window overlaps:
Performance windows may overlap when analyzing content over different time periods. This means the same data points may appear in multiple analysis windows, which can affect comparisons.

Conclusion:
This data can help identify possible content refresh opportunities, but it cannot prove the exact cause of performance changes or guarantee that refreshing content will improve results.
